```python
# ── When does mCherry plateau? Find the right truncation point ────────────
MIN_YIELD = 200  # ignore failed/near-zero experiments for this analysis

plateau_steps = []  # step at which each experiment plateaus (within 2% of final value)

for key, entry in manifest.items():
    o   = pd.read_parquet(DATA_DIR / entry["outputs_path"])
    mch = o["mCherry [RFU]"].to_numpy()
    final = mch[-1]
    if final < MIN_YIELD:
        continue  # skip failed experiments

    # find first step where mCherry is within 2% of final value and stays there
    threshold = 0.98 * final
    plateau_at = len(mch)  # default: never reaches plateau
    for k in range(len(mch) - 1, -1, -1):
        if mch[k] < threshold:
            plateau_at = k + 1
            break
    plateau_steps.append({"key": key, "K": len(mch),
                           "plateau_at": plateau_at, "final_mcherry": final})

df_p = pd.DataFrame(plateau_steps)
print(f"Experiments with yield > {MIN_YIELD}: {len(df_p)} / {len(manifest)}")
print(f"\nPlateau step distribution:")
print(df_p["plateau_at"].describe(percentiles=[.25,.5,.75,.9,.95,.99]).round(0))

# Plot distribution of plateau steps
fig, axes = plt.subplots(1, 2, figsize=(12, 3))
axes[0].hist(df_p["plateau_at"], bins=40, color="tomato", edgecolor="white")
axes[0].axvline(df_p["plateau_at"].quantile(0.95), color="k", ls="--",
                label=f"95th pct = {df_p['plateau_at'].quantile(0.95):.0f}")
axes[0].set(title="Step at which mCherry plateaus (98% of final)",
            xlabel="step", ylabel="count")
axes[0].legend()

axes[1].hist(df_p["plateau_at"] * 60 / 3600, bins=40, color="tomato", edgecolor="white")
axes[1].axvline(df_p["plateau_at"].quantile(0.95) * 60 / 3600, color="k", ls="--")
axes[1].set(title="Same in hours", xlabel="hours")
fig.tight_layout()
plt.show()

K95 = int(df_p["plateau_at"].quantile(0.95))
print(f"\nRecommended truncation: K={K95} ({K95*60/3600:.1f}h) covers 95% of experiments")
```

```python
# ── How many experiments survive different truncation strategies? ──────────
MIN_YIELD  = 200
K_TRUNC    = 1446  # 95th percentile plateau step

survive = {"key": [], "K": [], "final_mcherry": [], "kept": [], "drop_reason": []}

for key, entry in manifest.items():
    o     = pd.read_parquet(DATA_DIR / entry["outputs_path"])
    K     = len(o)
    final = o["mCherry [RFU]"].iloc[-1]

    if final < MIN_YIELD:
        reason = "low_yield"
        kept   = False
    elif K < K_TRUNC:
        reason = "too_short"
        kept   = False
    else:
        reason = ""
        kept   = True

    survive["key"].append(key)
    survive["K"].append(K)
    survive["final_mcherry"].append(final)
    survive["kept"].append(kept)
    survive["drop_reason"].append(reason)

df_s = pd.DataFrame(survive)
print(f"Total experiments:  {len(df_s)}")
print(f"Kept (K≥{K_TRUNC} and yield≥{MIN_YIELD}):  {df_s['kept'].sum()}")
print(f"Dropped — too short:  {(df_s['drop_reason']=='too_short').sum()}")
print(f"Dropped — low yield:  {(df_s['drop_reason']=='low_yield').sum()}")

# Are short experiments also low-yield, or are they a separate group?
short = df_s[df_s["drop_reason"] == "too_short"]
kept  = df_s[df_s["kept"]]
print(f"\nShort experiments — mCherry distribution:")
print(short["final_mcherry"].describe(percentiles=[.25,.5,.75]).round(1))
print(f"\nKept experiments — mCherry distribution:")
print(kept["final_mcherry"].describe(percentiles=[.25,.5,.75]).round(1))

# Are short experiments from specific date batches?
df_s["date"] = df_s["key"].str.extract(r"(\d{4}-\d{2}-\d{2})")
short_dates  = df_s[df_s["drop_reason"] == "too_short"]["date"].value_counts()
print(f"\nShort experiments by date batch (top 5):")
print(short_dates.head())
```

## Test-split seed survey

Vary `split_seed` over many values and check whether the resulting test set has a balanced coverage of protein yields and mRNA max. The default seed (42) can drop almost all high-yield experiments into the test set, making R² look worse than it is. Pick a seed where both the train and test histograms look similar.

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd

# ── Config — mirrors train_R.py exactly ──────────────────────────────────
NPZ_PATH         = Path("../../datasets/real_ivtt_raw_no_outliers.npz")
N_TEST           = 104          # ~15% of 695
N_VAL            = 104
STRATIFY_BINS    = 5
STRATIFY_TARGETS = [3, 5]       # mm=3, pm=5 in 7-state [R,O,m,mm,p,pm,DNA]
SEEDS            = list(range(200))

d = np.load(str(NPZ_PATH), allow_pickle=True)
y_seq       = d["y_seq"]
obs_names   = list(d["obs_names"].astype(str))
lengths_arr = d["lengths"].astype(np.int64) if "lengths" in d else None
N           = y_seq.shape[0]

pm_idx = obs_names.index("pm")
mm_idx = obs_names.index("mm")

# ── Exact copies of helpers from train_R.py ──────────────────────────────
def _endpoint_values(y_seq, lengths, target_idx):
    N = y_seq.shape[0]
    if lengths is None:
        return y_seq[:, -1, target_idx].astype(np.float64)
    end_t = np.clip(lengths.astype(np.int64) - 1, 0, y_seq.shape[1] - 1)
    out = np.empty((N, len(target_idx)), dtype=np.float64)
    for j, t_idx in enumerate(target_idx):
        out[:, j] = y_seq[np.arange(N), end_t, t_idx]
    return out

def _quantile_bin_1d(values, n_bins):
    if n_bins <= 1:
        return np.zeros(values.shape[0], dtype=np.int64)
    edges = np.unique(np.quantile(values, np.linspace(0.0, 1.0, n_bins + 1)))
    if edges.size <= 2:
        return np.zeros(values.shape[0], dtype=np.int64)
    return np.digitize(values, edges[1:-1], right=True).astype(np.int64)

def _allocate_counts(stratum_sizes, total):
    out = np.zeros_like(stratum_sizes, dtype=np.int64)
    if total <= 0 or stratum_sizes.sum() <= 0:
        return out
    raw  = total * (stratum_sizes / max(1, int(stratum_sizes.sum())))
    base = np.minimum(np.floor(raw).astype(np.int64), stratum_sizes)
    out[:] = base
    need = int(total - out.sum())
    for i in np.argsort(-(raw - np.floor(raw))):
        if need <= 0: break
        if out[i] < stratum_sizes[i]: out[i] += 1; need -= 1
    for i in np.argsort(-stratum_sizes):
        if need <= 0: break
        take = min(need, int(stratum_sizes[i] - out[i]))
        if take > 0: out[i] += take; need -= take
    return out

def make_splits(seed):
    rng  = np.random.default_rng(seed)
    # rank-average across stratify targets → quantile bins
    vals = _endpoint_values(y_seq, lengths_arr, STRATIFY_TARGETS)
    vals = vals[:, vals.std(axis=0) > 1e-12]
    ranks = np.empty_like(vals, dtype=np.float64)
    for j in range(vals.shape[1]):
        order = np.argsort(vals[:, j], kind="mergesort")
        inv = np.empty_like(order); inv[order] = np.arange(N)
        ranks[:, j] = inv / max(1, N - 1)
    labels = _quantile_bin_1d(ranks.mean(axis=1), STRATIFY_BINS)

    strata = []
    for u in np.unique(labels):
        idx_u = np.where(labels == u)[0]; rng.shuffle(idx_u); strata.append(idx_u)

    sizes     = np.array([len(s) for s in strata], dtype=np.int64)
    take_test = _allocate_counts(sizes, N_TEST)
    test_parts, rem_parts = [], []
    for s, k in zip(strata, take_test):
        test_parts.append(s[:k]); rem_parts.append(s[k:])

    rem_sizes = np.array([len(s) for s in rem_parts], dtype=np.int64)
    take_val  = _allocate_counts(rem_sizes, N_VAL)
    val_parts, train_parts = [], []
    for s, k in zip(rem_parts, take_val):
        val_parts.append(s[:k]); train_parts.append(s[k:])

    test_idx  = np.concatenate(test_parts);  rng.shuffle(test_idx)
    val_idx   = np.concatenate(val_parts);   rng.shuffle(val_idx)
    train_idx = np.concatenate(train_parts); rng.shuffle(train_idx)
    return train_idx, val_idx, test_idx

# ── Endpoint values for scoring ──────────────────────────────────────────
protein_final = _endpoint_values(y_seq, lengths_arr, [pm_idx])[:, 0]
mrna_max_vals = (y_seq[:, :, mm_idx].max(axis=1) if lengths_arr is None else
                 np.array([y_seq[i, :int(lengths_arr[i]), mm_idx].max() for i in range(N)]))

ref_pf_mean = protein_final.mean();  pf_std = protein_final.std() + 1e-9
ref_mm_mean = mrna_max_vals.mean();  mm_std = mrna_max_vals.std()  + 1e-9

# ── Survey all seeds ──────────────────────────────────────────────────────
records = []
for s in SEEDS:
    tr, va, te = make_splits(s)
    z2, row = 0.0, {"seed": s}
    for name, idx in [("train", tr), ("val", va), ("test", te)]:
        pf = protein_final[idx]; mm = mrna_max_vals[idx]
        row[f"{name}_pf_mean"]   = pf.mean()
        row[f"{name}_pf_max"]    = pf.max()
        row[f"{name}_pf_n_gt2k"] = int((pf > 2000).sum())
        row[f"{name}_mm_mean"]   = mm.mean()
        z2 += ((pf.mean() - ref_pf_mean) / pf_std) ** 2
        z2 += ((mm.mean() - ref_mm_mean) / mm_std) ** 2
    row["score"] = z2 ** 0.5
    records.append(row)

df = pd.DataFrame(records).set_index("seed")

print(f"N={N}  n_test={N_TEST} ({N_TEST/N*100:.0f}%)  n_val={N_VAL} ({N_VAL/N*100:.0f}%)  "
      f"n_train={N-N_TEST-N_VAL}")
print(f"Stratified split: targets={STRATIFY_TARGETS} ({[obs_names[i] for i in STRATIFY_TARGETS]}), "
      f"bins={STRATIFY_BINS}")
print(f"Global protein final: mean={ref_pf_mean:.1f}  max={protein_final.max():.1f}")
print(f"Global mRNA max:      mean={ref_mm_mean:.1f}  max={mrna_max_vals.max():.1f}")
print()
cols = ["train_pf_mean","val_pf_mean","test_pf_mean",
        "train_mm_mean","val_mm_mean","test_mm_mean","test_pf_n_gt2k","score"]
print("Top 15 balanced seeds (stratified, all three splits):")
print(df.sort_values("score").head(15)[cols].round(1).to_string())
```

## Sanity check: which obs channels in `real_ivtt_full.npz` actually carry signal?

The dataset declares 6 observed species (`R, m, mm, p, pm, DNA`) but the raw experiments only measure
**Broccoli (→ mm)** and **mCherry (→ pm)**. The other 4 channels were padded with zeros by the converter,
yet downstream code (e.g. `per_step_theta_fit.py`) treats them as real observations and includes them in
the loss — which forces the ODE to crush latent species to zero and breaks rollout.

The cells below verify this directly across many samples.

```python
# ── Load the .npz the fitter actually consumes ───────────────────────────
import numpy as np

NPZ_PATH = Path("../../datasets/real_ivtt_raw_no_outliers.npz")
npz = np.load(NPZ_PATH, allow_pickle=True)

obs_names = list(npz["obs_names"].astype(str))
y_seq     = npz["y_seq"]   # (N, K, P)
y0        = npz["y0"]      # (N, P)
N, K, P   = y_seq.shape
print(f"y_seq shape: {y_seq.shape}  (N={N} samples, K={K} steps, P={P} channels)")
print(f"obs_names:   {obs_names}")
```

```python
# ── Per-channel summary across ALL 696 samples ───────────────────────────
rows = []
for i, name in enumerate(obs_names):
    ch = y_seq[:, :, i]                                # (N, K)
    n_all_zero = int((ch == 0).all(axis=1).sum())      # samples where ch is 0 at every step
    frac_nz    = float((ch != 0).mean())               # fraction of (N*K) entries that are nonzero
    rows.append({
        "channel":          name,
        "samples_all_zero": f"{n_all_zero}/{N}",
        "global_min":       float(ch.min()),
        "global_max":       float(ch.max()),
        "global_mean":      float(ch.mean()),
        "frac_entries_nz":  f"{frac_nz:.3%}",
    })

summary = pd.DataFrame(rows)
print(summary.to_string(index=False))
```
